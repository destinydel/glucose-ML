```python
import os
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
cgm_data = pd.read_csv('C:/Users/Desti/OneDrive/Desktop/glucose-ML/AZ1TD/data/CGM Records/Subject 1/Subject 1.csv')


cgm_data.info() #some columns have missing values, but will keep for now
```

```python
import matplotlib.dates as mdates

plt.figure(figsize=(12, 6))

plt.plot(cgm_data['EventDateTime'], cgm_data['CGM'])

# Show a time label every hour
plt.gca().xaxis.set_major_locator(mdates.HourLocator(interval=1))
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%H:%M'))

plt.xticks(rotation=45)
plt.tight_layout()

plt.xlabel('Time')
plt.ylabel('Glucose')
plt.title('CGM Readings Over Time')

plt.show()
```
